```python
# 11. Write a function to withdraw money and raise an exception if balance < withdrawal.
def withdrawal_exception(balance, withdrawl):
    if balance < withdrawl:
        raise ValueError("Insufficient balance to withdraw")
    return balance - withdrawl
try:
    current_balance = int(input("Enter the balance : "))
    withdrawal_amount = int(input("Enter the withdrawal : "))
    new_balance = withdrawal_exception(current_balance, withdrawal_amount)
    print("Withdrawal successful. New balance:", new_balance)
except ValueError as e:
    print("Transaction failed:", e)

```

```python
# 12. Write a function that validates age and raises an exception if age < 18.
def voter(age):
    if age < 18:
        raise Exception(" Candidate is not eligible to Vote ")
    return age
try:
    age=int(input("Enter the age: "))
    eligible_age=voter(age)
    print("Candidate is eligible to vote: ",eligible_age)
except Exception as e:
    print("Error: ",e)
```

```python
# 14. Write a custom exception NegativeNumberError and use it in a square root function.
import math
class NegativeNumberError(Exception):
    pass
def square_root(n):
    try:
        if n < 0:
            raise NegativeNumberError("Cannot calculate square root of a negative number")
        else:
            result = math.sqrt(n)
            print(f"Square root of {n} is {result}")
    except Exception as e:
        print("Error:", e)
num = int(input("Enter a number: "))
square_root(num)

```

```python
# 15. Write a custom exception for InsufficientFundsError in a banking system.
class InsufficientFundsError(Exception):
    pass
def withdrawal_exception(balance, withdrawal):
    if balance < withdrawal:
        raise InsufficientFundsError("Insufficient balance to withdraw")
    return balance - withdrawal
try:
    current_balance = int(input("Enter your balance: "))
    withdrawal_amount = int(input("Enter the withdrawal amount: "))
    new_balance = withdrawal_exception(current_balance, withdrawal_amount)
    print("Withdrawal successful. New balance:", new_balance)
except InsufficientFundsError as e:
    print("Transaction failed:", e)
```

```python
# 17. Write a program to retry dividing two numbers 3 times if an error occurs.
def divide(a, b, times=3):
    for attempt in range(1, times + 1):
        try:
            result = a / b
            print(f"Division successful on attempt {attempt}: {result}")
            return result
        except ZeroDivisionError:
            print(f"Attempt {attempt}: Cannot divide by zero.")
    print(f"Failed after {times} attempts.")
a=int(input("enter the first number"))
b=int(input("enter the second number"))
divide(a,b)
```

```python
# 18. Write a program that uses assert to validate positive numbers and handle AssertionError.
try:
    num=int(input("enter a number"))
    assert num > 0 ,"Number must be Positive"
except AssertionError as e:
    print("Error!",e)
except ValueError:
    print("Invalid input! Please enter an integer.")
else:
    print("Number is positive")
```
